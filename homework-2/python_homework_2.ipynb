# Задание 1

Печатные газеты использовали свой формат дат для каждого выпуска. Для каждой газеты из списка напишите формат указанной даты для перевода в объект datetime:

The Moscow Times - Wednesday, October 2, 2002

The Guardian - Friday, 11.10.13

Daily News - Thursday, 18 August 1977

Пример работы программы

Программа должна выводить на экран объекты типа datetime, соответствующие датам в условии задачи

```python
from datetime import datetime

newspaper_dict = {
    "The Moscow Times": ["Wednesday, October 2, 2002", "%A, %B %d, %Y"],
    "The Guardian": ["Friday, 11.10.13", "%A, %d.%m.%y"],
    "Daily News": ["Thursday, 18 August 1977", "%A, %d %B %Y"]
}

for newspaper, (str_date, dt_format) in newspaper_dict.items():

    datetime_date = datetime.strptime(str_date, dt_format)
    print(newspaper, "-", datetime_date)
```

```python
#доп. решение

from datetime import datetime

user_input = input()

name_date = [element.strip() for element in user_input.split("-")]

newspaper = name_date[0]
date = name_date[1]

if newspaper == "The Moscow Times":
    date_datetime = datetime.strptime(date, '%A, %B %d, %Y')
    
    print(date_datetime, type(date_datetime))
elif newspaper == "The Guardian":
    date_datetime = datetime.strptime(date, '%A, %d.%m.%y')
    
    print(date_datetime, type(date_datetime))
elif newspaper == "Daily News":
    date_datetime = datetime.strptime(date, '%A, %d %B %Y')
    
    print(date_datetime, type(date_datetime))
```


# Задание 2 (не обязательное)

Напишите функцию date_range, которая возвращает список дат за период от start_date до end_date. Даты должны вводиться в формате YYYY-MM-DD. В случае неверного формата или при start_date > end_date должен возвращаться пустой список.

 

Примеры работы программы:

date_range(‘2022-01-01’, ‘2022-01-03’)

[‘2022-01-01’, ‘2022-01-02’, ‘2022-01-03’]

```python
from datetime import datetime, timedelta


def date_range(start_date, end_date):

    dates_list = []

    try:
        start_date_datetime = datetime.strptime(start_date, "%Y-%m-%d")
        end_date_datetime = datetime.strptime(end_date, "%Y-%m-%d")
        
    except ValueError:
        return dates_list

    if start_date_datetime > end_date_datetime:
        return dates_list

    while start_date_datetime <= end_date_datetime:
        dates_list.append(datetime.strftime(start_date_datetime, "%Y-%m-%d"))
        
        if start_date_datetime == end_date_datetime:
            break
            
        start_date_datetime += timedelta(days=1)
            
        
    return dates_list
```

```python
date_range('2022-01-01', '2022-01-03')
```

```python
date_range('2022-01-03', '2022-01-01')
```

```python
date_range('2022-02-30', '2022-02-31')
```
